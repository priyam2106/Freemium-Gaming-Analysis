# **DAYS SINCE LAST PURCHASE**

```python
df['Last Purchase Date']=pd.to_datetime(df['LastPurchaseDate'])
df['Days Since Last Purchase']=(pd.Timestamp.today()-df['Last Purchase Date']).dt.days
```

```python
df['Days Since Last Purchase'].sort_values(ascending=False)
```

# **CREATE PURCHASE SPEED**

```python
df['Purchase Speed']=np.where(df['FirstPurchaseDaysAfterInstall']<=7,'Fast','Slow')
```

```python
df['Purchase Speed']
```

```python
df['Purchase Speed'].value_counts()
```

```python
sns.countplot(x='Purchase Speed',data=df)
plt.title("Purchase Speed")
plt.show()
```

# **CREATE AGE GROUPS**

```python
bins = [0,18,25,35,45,100]

labels = ['Teen','Young Adult','Adult','Middle Age','Senior']
df['Age Group']=pd.cut(df['Age'],bins=bins,labels=labels)
```

```python
df['Age Group'].value_counts()
```

# **CREATE SESSION CATEGORIES**

```python
df['Session Category']=pd.cut(df['SessionCount'],bins=[0,20,50,100,500],
                              labels=['Low','Medium','High','Very High'])
```

```python
df[df["Session Category"]== "Medium"]
```

```python
df['Session Category'].value_counts()
```

# **CREATE SESSION LENGTH CATEGORIES**

```python
df['Session Length Category']=pd.cut(df['AverageSessionLength'],bins=[0,10,20,40,100],labels=['Short','Medium','Long','Very long'])
```

```python
df
```

```python
df['Session Length Category'].value_counts()
```

```python
sns.countplot(x='Session Length Category',data=df)
plt.title("Session Length Category")
plt.show()
```

# **TOTAL PLAYERS**

# **TOTAL REVENUE**

# **AVERAGE REVENUE PER USER**

```python
arpu = total_revenue/total_players
print(f"ARPU : ${arpu:.2f}")
```

# **AVERAGE PURCHASE AMOUNT**

```python
avg_purchase = df['InAppPurchaseAmount'].mean()
print(f"Average Purchase Amount: ${avg_purchase:.2f}")
```

# **MEDIUM PURCHASE AMOUNT**

```python
median_purchase = df['InAppPurchaseAmount'].median()
print(f"Median Purchase: ${median_purchase:.2f}")
```

# **HIGHEST SINGLE PURCHASE**

```python
highest_purchase = df['InAppPurchaseAmount'].max()
print(f"Highest Single Purchase: ${highest_purchase:.2f}")
```
